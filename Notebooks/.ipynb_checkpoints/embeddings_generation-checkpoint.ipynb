```python
import sys
from unittest.mock import MagicMock
sys.modules['pyBigWig'] = MagicMock()
import pandas as pd
import numpy as np
import torch
import gpn.model
import torch.nn.functional as F
from transformers import AutoModel, AutoModelForMaskedLM
from gpn.data import GenomeMSA  # should work now
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
```

```python
coding = pd.read_csv('/home/apatil2/Disease_risk_modeling/filtered_data_link/clinvar_20240805.missense_matched.txt',
                sep = '\t',
                low_memory = False)
```

```python
#loading the MSA file
msa_path = "/blue/egn6933/apatil2/pz/89.zarr"  # ← plain folder
genome_msa = GenomeMSA(msa_path)
```

```python
#loading the model
model_path = 'songlab/gpn-msa-sapiens'
model = AutoModel.from_pretrained(model_path).to(device)
model.eval()
```

```python
#function to get the embeddings
def get_genome_embedding(chrom, pos_start, pos_end, model):
    
    msa = genome_msa.get_msa(str(chrom), pos_start, pos_end, strand="+", tokenize=True)
    msa = torch.tensor(np.expand_dims(msa, 0).astype(np.int64))
    
    #separating human from rest of species
    input_ids, aux_features = msa[:, :, 0], msa[:, :, 1:]
    input_ids.shape, aux_features.shape
    
    #pushing data to gpu
    input_ids = input_ids.to(device)
    aux_features = aux_features.to(device)
    
    with torch.no_grad():
        embeddings = model(input_ids=input_ids, aux_features=aux_features).last_hidden_state
        
        return embeddings
```

## Embedding retrieval approaches comparison

```python
#first approach 
e = get_genome_embedding(2, coding['POS'][0], coding['POS'][0]+1, model)

#second approach
e1 = get_genome_embedding(2, coding['POS'][0]-64, coding['POS'][0]+64, model)
```

```python
# e.shape
# reshape the matrices
e = e.view(1,768)
e1 = e1.view(128, 768)
```

```python
# first approach embeddings
e2 = get_genome_embedding(2, coding['POS'][1], coding['POS'][1]+1, model)
e2 = e2.view(1, 768)

# second approach embeddings
e3 = get_genome_embedding(2, coding['POS'][1]-64, coding['POS'][1]+64, model)
e3 = e3.view(128, 768)
```

```python
# compare the similarity of the embeddings
sim = F.cosine_similarity(e[0], e1[64], dim=0)
print(sim)
```

```python
sim1 = F.cosine_similarity(e2[0], e3[64], dim=0)
print(sim1)
```

## Combining the two dataframes

```python
noncoding = pd.read_csv('/home/apatil2/Disease_risk_modeling/filtered_data_link/data/clinvar_20240805.noncoding.txt',
                sep = '\t',
                low_memory = False)

coding = pd.read_csv('/home/apatil2/Disease_risk_modeling/filtered_data_link/clinvar_20240805.missense_matched.txt',
                sep = '\t',
                low_memory = False)
```

```python
coding.columns
```

```python
noncoding.columns
```

```python
#filtering the unwanted columns
coding_small = coding[['CHROM', 'POS', 'REF', 'ALT', 'Pathogenicity']]
noncoding_small = noncoding[['CHROM', 'POS', 'REF', 'ALT', 'Pathogenicity']]
```

```python
# combining the dataframes
combined = pd.concat([coding_small, noncoding_small], axis = 0, ignore_index = True)

# mixing the rows to mix the coding and noncoding variants
combined = combined.sample(frac=1, random_state=42).reset_index(drop=True)

combined['Pathogenicity'] = combined['Pathogenicity'].map({'B': 0, 'P': 1})
```

```python
#saving the combined df
combined.to_csv('combined_df.csv', index=False)
```

## Saving the embeddings 

```python
def convert_embeddings_to_torch(batch_size = 256):
    embedding_list = []
    total = len(combined)
    
    for start in range(0, total, batch_size):
        end = min(start + batch_size, total)
        
        batch_embeddings = []
        batch_labels = []
        
        for i in range(start, end):
            row = combined.iloc[i]
            
            result = get_genome_embedding(
                row['CHROM'],
                row['POS'] - 64,
                row['POS'] + 64,
                model
            )
            
            result = result.view(128, 768).detach().cpu()
            
            batch_embeddings.append(result)
            batch_labels.append(row['Pathogenicity'])
        
        X_batch = torch.stack(batch_embeddings)
        y_batch = torch.tensor(batch_labels)
        
        torch.save(
            {'X': X_batch, 'Y': y_batch},
            f'/blue/egn6933/apatil2/embeddings/embeddings_batch_{start}.pt'
        )
        
        print(f'Saved batch {start} to {end}')
        
        del batch_embeddings, batch_labels, X_batch, y_batch
        torch.cuda.empty_cache()
```

```python
convert_embeddings_to_torch()
```
