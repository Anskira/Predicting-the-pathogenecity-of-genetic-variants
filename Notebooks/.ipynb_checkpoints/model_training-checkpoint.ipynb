```python
import pandas as pd
import torch
import transformers
from torch.utils.data import Dataset
import os
from torch.utils.data import DataLoader
```

```python
# define the dataset class
class EmbeddingDataset(Dataset):
    def __init__(self, folder):
        self.folder = folder
        self.files = sorted(os.listdir(folder))
    
    
    def __len__(self):
        
        return len(self.files)
    
    def __get__(self, idx):
        batch = torch.load(os.path.join(self.folder, self.files[idx]))
        
        X = batch['X']
        y = batch['Y']
        
        return X, y                        
```

```python
# define the data loader

dataset = EmbeddingDataset('/blue/egn6933/apatil2/embeddings/')

loader = DataLoader(dataset, batch_size = 1, shuffle = False)
```

```python
import torch.nn as nn

class MLP(nn.module):
    
    def __init__(self):
        super().__init__()
    
        self.model = nn.Sequential(
            nn.Linear(768, 512),
            nn.ReLU(),
            nn.Dropout(0.3),
            
            nn.Linear(512, 256),
            nn.ReLU(),
            
            nn.Linear(256, 1)
        )
    
    def forward(self, x):
        return self.model(x)
    
```

```python
#training configuration

device = 'cuda'

model = MLP().to(device)

criterion = nn.BCEWithLogitsLoss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr = 1e-4
)
```
